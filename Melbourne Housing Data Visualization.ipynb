```python
#Interactive Visualization of Melbourne Housing Data

import pandas as pd           
import matplotlib.pyplot as plt  


df = pd.read_csv("C:\\Users\\DELL\\Downloads\\melb_data.csv\\melb_data.csv")  

plt.figure(figsize=(8, 4))

plt.scatter(df['Distance'], df['Price'], alpha=0.5) 

plt.title("Static Scatter Plot: Distance vs Price")
plt.xlabel("Distance from CBD (km)")
plt.ylabel("Price (AUD)")
plt.grid(True)


plt.show()
```

```python
import plotly.express as px 
fig = px.scatter(df,
                 x='Distance',         
                 y='Price',            
                 color='Type',         
                 hover_data=['Suburb', 'Rooms'],  
                 title="Interactive Scatter Plot: Distance vs Price")

fig.show()
```

```python

plt.figure(figsize=(10, 6))
df.boxplot(column='Price', by='Type') 


plt.title("Static Box Plot: Price Distribution by Property Type")
plt.suptitle("")  
plt.xlabel("Property Type")
plt.ylabel("Price (AUD)")


plt.show()
```

```python
fig = px.box(df,
             x='Type',          
             y='Price',          
             color='Type',      
             title="Interactive Box Plot: Price Distribution by Property Type",
             points="all")       

fig.show()
```

```python
plt.figure(figsize=(10, 4))
plt.hist(df['Price'], bins=30, color='skyblue', edgecolor='black')  


plt.title("Static Histogram: Distribution of Property Prices")
plt.xlabel("Price (AUD)")
plt.ylabel("Frequency")


plt.show()
```
